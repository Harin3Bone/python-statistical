<font color="#275bb0">**Read dataset**</span>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import metrics
```

```python
df_agg = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Aggregation.csv")
df_dens = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Density.csv")
df_flame = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Flame.csv")
df_nclus = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Nclus.csv")
df_spiral = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Spiral.csv")
df_toy = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/cluster_datasets/Toy.csv")
```

&nbsp;  
<font color="#275bb0">**Visualize dataset**</span>

```python
fig, ax = plt.subplots(2, 3, figsize=(15, 10))

sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[0,0]).set(title='Aggregation')

sns.scatterplot(data=df_dens,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[0,1]).set(title='Density')

sns.scatterplot(data=df_flame,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[0,2]).set(title='Flame')

sns.scatterplot(data=df_nclus,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[1,0]).set(title='Nclus')

sns.scatterplot(data=df_spiral,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[1,1]).set(title='spiral')

sns.scatterplot(data=df_toy,
                x='x',
                y='y',
                alpha=.5,
                ax=ax[1,2]).set(title='Toy')
```

&nbsp;  
#### <font color="#166e20">*K-Means Clustering*</span>

&nbsp;  
<font color="#275bb0">**Configure k-means parameters**</span>

```python
from sklearn.cluster import KMeans
km = KMeans(n_clusters=7, # The number of clusters to form as well as the number of centroids to generate. default=8
            random_state=1234) # Determines random number generation for centroid initialization.
```

&nbsp;  
<font color="#275bb0">**Class prediction using k-means clustering**</span>

- Aggregation dataset

```python
y_pred = km.fit_predict(df_agg)
y_pred
```

&nbsp;  
<font color="#275bb0">**Visualize clustered classes**</span>

```python
sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred).set(title='Aggregation')
```

&nbsp;  
<font color="#275bb0">**Using k=2-10**</span>

```python
fig, ax = plt.subplots(2, 5, figsize=(15, 6))

i=0
for k in range(2,7):
    km = KMeans(n_clusters=k, random_state=1234)
    y_pred = km.fit_predict(df_agg)
    sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                ax=ax[0,i]).set(title=f'k={k}')
    i+=1

i=0
for k in range(7,11):
    km = KMeans(n_clusters=k, random_state=1234)
    y_pred = km.fit_predict(df_agg)
    sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                ax=ax[1,i]).set(title=f'k={k}')
    i+=1
```

Well, we don't have a (true/actual) label.  
Which k should we used?

&nbsp;  
<font color="#275bb0">**Selecting k**</span>

We can get **within-cluster sum-of-squares** using attribute .inertia_:

```python
# When using k=2
km = KMeans(n_clusters=2, random_state=1234)
y_pred = km.fit_predict(df_agg)
km.inertia_
```

```python
in_ss = {'k':[], 'within-ss':[]}
for k in range(2,11):
    km = KMeans(n_clusters=k, random_state=1234)
    y_pred = km.fit_predict(df_agg)
    in_ss['k'].append(k)
    in_ss['within-ss'].append(km.inertia_)

in_ss = pd.DataFrame(in_ss)
sns.lineplot(data=in_ss, x='k', y='within-ss', marker='o')
```

\
Let's try with 'Density' dataset:

```python
fig, ax = plt.subplots(2, 5, figsize=(20, 9))
in_ss = {'k':[], 'within-ss':[]}

i=0
for k in range(2,7):
    km = KMeans(n_clusters=k, random_state=1234)
    y_pred = km.fit_predict(df_dens)
    sns.scatterplot(data=df_dens,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                s=30,
                ax=ax[0,i]).set(title=f'k={k}')
    i+=1
    in_ss['k'].append(k)
    in_ss['within-ss'].append(km.inertia_)

i=0
for k in range(7,11):
    km = KMeans(n_clusters=k, random_state=1234)
    y_pred = km.fit_predict(df_dens)
    sns.scatterplot(data=df_dens,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                s=30,
                ax=ax[1,i]).set(title=f'k={k}')
    i+=1
    in_ss['k'].append(k)
    in_ss['within-ss'].append(km.inertia_)
```

```python
in_ss = pd.DataFrame(in_ss)
sns.lineplot(data=in_ss, x='k', y='within-ss', marker='o')
```

&nbsp;  
#### <font color="#166e20">*K-Means Clustering Using a Dataset With Known Actual Class Label*</span>

We can also perform clustering on dataset that true label already provided to evaluate 'clustering performance'.  
For example, let's try clustering on 'Wholesale_customers_data.csv'.

```python
df = pd.read_csv("https://raw.githubusercontent.com/ThammakornS/ProgStat/main/dataset/iris.csv")
df.head()
```

\
*For demonstration, we will use **only petal-length and sepal-length** data.*

```python
df = df.filter(['petal-length', 'sepal-length', 'Class'])
```

```python
sns.scatterplot(data=df,
                x='petal-length',
                y='sepal-length',
                hue='Class',
                alpha=.7)
```

&nbsp;  
<font color="#275bb0">**Create feature and label data**</span>

\
Feature data:

```python
X = df.loc[:,~(df.columns=='Class')]
X
```

\
Label data:

```python
y = df['Class']
y
```

\
Perform clustering using k=3:

```python
km = KMeans(n_clusters=3, random_state=1234)
y_pred = km.fit_predict(X)
y_pred
```

&nbsp;  
<font color="#275bb0">**Visualize clustering result**</span>

```python
fig, ax = plt.subplots(figsize=(10, 10))
sns.scatterplot(data=df,
                x='petal-length',
                y='sepal-length',
                hue='Class',
                s=200,
                alpha=.7,
                ax=ax)
sns.scatterplot(data=df,
                x='petal-length',
                y='sepal-length',
                hue=y_pred,
                palette=['red', 'purple', 'cyan'],
                marker="^",
                s=60,
                alpha=.7,
                ax=ax).set(title="Circle:Actual Class | Triangle:Clustered Class")
```

&nbsp;  
<font color="#275bb0">**Evaluate performance scores**</span>

\
Then we can evaluate clustering performance scores:

see https://scikit-learn.org/stable/modules/clustering.html#clustering-performance-evaluation

```python
print("Homogeneity: %0.3f" % metrics.homogeneity_score(y, y_pred))
print("Completeness: %0.3f" % metrics.completeness_score(y, y_pred))
print("V-measure: %0.3f" % metrics.v_measure_score(y, y_pred))
print("Rand Index: %0.3f" % metrics.rand_score(y, y_pred))
print("Adjusted Rand Index: %0.3f" % metrics.adjusted_rand_score(y, y_pred))
print("Mutual Information: %0.3f" % metrics.mutual_info_score(y, y_pred))
print("Adjusted Mutual Information: %0.3f" % metrics.adjusted_mutual_info_score(y, y_pred))
```

Note: maximum = 1  for all scores

&nbsp;  
#### <font color="#166e20">*Spectral Clustering*</span>

&nbsp;  
<font color="#275bb0">**Configure spectral clustering parameters**</span>

```python
from sklearn.cluster import SpectralClustering
spectral = SpectralClustering(n_clusters=7, # The expected number of cluster (default=8)
                              affinity='rbf', # Kernel to construct the affinity matrix
                              random_state=1234) # A pseudo random number generator used for the initialization of the lobpcg eigenvectors decomposition
```

&nbsp;  
<font color="#275bb0">**Class prediction using spectral clustering**</span>

- Aggregation dataset

```python
y_pred = spectral.fit_predict(df_agg)
y_pred
```

&nbsp;  
<font color="#275bb0">**Visualize clustered classes**</span>

```python
sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred).set(title='Density')
```

&nbsp;  
<font color="#275bb0">**Using n_clusters from 2 to 10**</span>

Note: SpectralClustering() not provide **within-cluster sum-of-squares**:

```python
fig, ax = plt.subplots(2, 5, figsize=(20, 9))

i=0
for n in range(2,7):
    spectral = SpectralClustering(n_clusters=n)
    y_pred = spectral.fit_predict(df_agg)
    sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                s=30,
                ax=ax[0,i]).set(title=f'n_clusters={n}')
    i+=1

i=0
for n in range(7,11):
    spectral = SpectralClustering(n_clusters=n)
    y_pred = spectral.fit_predict(df_agg)
    sns.scatterplot(data=df_agg,
                x='x',
                y='y',
                alpha=.5,
                c=y_pred,
                s=30,
                ax=ax[1,i]).set(title=f'n_clusters={n}')
    i+=1
```
